Q1. Division with Exception Handling

Write a program that takes two numbers and safely performs division.
Handle division by zero using try-except.

````python
def num_devide(num1, num2):
    try:
        return num1/num2
    except ZeroDivisionError:
        print(f"Denominator can not be zero.")        

num_devide(5,0)    
````

Q2. String to Integer Conversion

Given a list:

data = ["10", "20", "abc", "30"]

Convert all values to integers.
Skip invalid values using exception handling.

````python
def type_conversion(my_list):
    for index, item in enumerate(my_list):
        try:
            my_list[index] = int(item) 
        except ValueError:
            continue
    return my_list

data = ["10","20","abc","30"]
print(type_conversion(data))

````

Q3. Multiple Exception Handling

Write a program that:

Takes user input

Divides 100 by that number

Handle:

ValueError
ZeroDivisionError

Q10. Reverse Words

Reverse words in a string:

"I love AI"

````python
def reverse_inplace(data):
    chr_lsit = list(data)
    i = 0
    e = len(chr_lsit) - 1

    while i < e:
        chr_lsit[i], chr_lsit[e] = chr_lsit[e], chr_lsit[i]
        i += 1
        e -= 1
    return "".join(chr_lsit)    

data = "I love AI"
print(reverse_inplace(data))
````
